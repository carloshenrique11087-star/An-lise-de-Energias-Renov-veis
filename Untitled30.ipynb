```python

def preparar_analise(df):
    df = df.copy()

    df.columns = df.columns.astype(str).str.strip()

    coluna_ano = next(
        (c for c in df.columns if c.lower() in ["year", "ano"]),
        None
    )

    if coluna_ano is None:
        if "Year" in df.index.names:
            df = df.reset_index()
            coluna_ano = "Year"
        elif "Ano" in df.index.names:
            df = df.reset_index()
            coluna_ano = "Ano"

    if coluna_ano is not None:
        df[coluna_ano] = pd.to_numeric(
            df[coluna_ano], errors="coerce"
        )

        df = df[df[coluna_ano].between(2000, 2023)]

        df = df.sort_values(coluna_ano)

    return df

mundo = preparar_analise(mundo)
regioes = preparar_analise(regioes)
america_latina = preparar_analise(america_latina)

print("Dados preparados para análise!")
```

```python


consumo_mensal = 300

potencia_solar = 3.0

horas_sol = 5

eficiencia = 0.80

tarifa = 0.90

investimento = 12000

percentual_autoconsumo = 0.35

fator_compensacao = 0.75

custo_fixo = 40

manutencao_anual = 0.01

print("DADOS DA SIMULAÇÃO")

print("Consumo mensal:", consumo_mensal, "kWh")
print("Potência solar:", potencia_solar, "kWp")
print("Horas de sol pleno:", horas_sol)
print("Tarifa:", tarifa, "R$/kWh")
print("Investimento:", investimento, "R$")
print("Autoconsumo direto:", percentual_autoconsumo * 100, "%")
print("Fator de compensação:", fator_compensacao * 100, "%")
print("Custo fixo mensal:", custo_fixo, "R$")
```

```python


dias_mes = 30

geracao_mensal = (
    potencia_solar *
    horas_sol *
    dias_mes *
    eficiencia
)

energia_consumida_direto = min(
    consumo_mensal,
    geracao_mensal * percentual_autoconsumo
)

energia_injetada = max(
    geracao_mensal - energia_consumida_direto,
    0
)

print(f"Geração mensal: {geracao_mensal:.2f} kWh")
print(f"Consumo direto: {energia_consumida_direto:.2f} kWh")
print(f"Energia injetada: {energia_injetada:.2f} kWh")
```

```python


energia_restante = max(
    consumo_mensal - energia_consumida_direto,
    0
)

creditos_utilizados = min(
    energia_injetada,
    energia_restante
)

economia_direta = (
    energia_consumida_direto * tarifa
)

economia_creditos = (
    creditos_utilizados *
    tarifa *
    fator_compensacao
)

conta_sem_solar = (
    consumo_mensal * tarifa + custo_fixo
)

economia_bruta = (
    economia_direta + economia_creditos
)

conta_com_solar = max(
    conta_sem_solar - economia_bruta,
    custo_fixo
)

economia_mensal = (
    conta_sem_solar - conta_com_solar
)

print(f"Conta sem solar: R$ {conta_sem_solar:.2f}")
print(f"Economia pelo consumo direto: R$ {economia_direta:.2f}")
print(f"Economia pelos créditos: R$ {economia_creditos:.2f}")
print(f"Conta com solar: R$ {conta_com_solar:.2f}")
print(f"Economia mensal: R$ {economia_mensal:.2f}")
```

```python


custo_manutencao_mensal = (
    investimento * manutencao_anual / 12
)

economia_liquida = (
    economia_mensal - custo_manutencao_mensal
)

if economia_liquida > 0:

    retorno_meses = investimento / economia_liquida

    retorno_anos = retorno_meses / 12

    print(f"Manutenção mensal estimada: R$ {custo_manutencao_mensal:.2f}")

    print(f"Economia líquida mensal: R$ {economia_liquida:.2f}")

    print(f"Retorno do investimento: {retorno_meses:.1f} meses")

    print(f"Retorno do investimento: {retorno_anos:.2f} anos")

else:
    print("Não existe economia líquida positiva para calcular o retorno.")
```

```python


categorias = [
    "Sem energia solar",
    "Com energia solar"
]

valores = [
    conta_sem_solar,
    conta_com_solar
]

plt.figure(figsize=(9, 5))

plt.bar(categorias, valores)

plt.title("Comparação da conta de energia")
plt.ylabel("Valor mensal (R$)")

plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
historico_ia = []

def perguntar_sers(pergunta):

    historico_texto = ""

    for mensagem in historico_ia:
        historico_texto += f"\n{mensagem['role']}: {mensagem['content']}"

    prompt = f"""
    {contexto_sers}

    HISTÓRICO DA CONVERSA:
    {historico_texto}

    PERGUNTA DO USUÁRIO:
    {pergunta}

    RESPONDA:
    """

    resposta = cliente_ia.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=prompt
    )

    texto_resposta = resposta.text

    historico_ia.append({
        "role": "Usuário",
        "content": pergunta
    })

    historico_ia.append({
        "role": "IA",
        "content": texto_resposta
    })

    return texto_resposta
```
